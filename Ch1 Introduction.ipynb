# Chapter 1: Introduction

### Why a new robotics book?

Most existing robotics textbooks are purely theoretical. Can we have a book that blends rigor with practice?

This online, interactive book tries to do exactly that. On one hand, we want to retain the rigor expected of a robotics fundamentals text. On the other hand, we want to show that the theory can be applied directly, both in simulation and on real hardware.

We start with the same theoretical foundations as traditional robotics textbooks. But we also include numerical implementations in Python to solve problems such as forward and inverse kinematics, the Jacobian, trajectory generation, and dynamics and control, to name a few.

Beyond these numerical implementations, we use MuJoCo to illustrate more complex robotic systems, simulating the kinematics and dynamics of open-source robot models directly.

Finally, we present two levels of hardware for implementing the theory and simulation:

- **Level 1 — Kinematics.** This level relies on smart bus servo motors that can directly control angle or speed. We will either build a simple robotic arm, such as a planar 2R arm, or use an existing low-cost, open-source arm like the SO-ARM101.
- **Level 2 — Dynamics and control.** This level is based on quasi-direct-drive motors, which can be controlled directly by torque. These can also be used to build a planar 2R robot, but they let us implement dynamics-based controllers, such as computed-torque control.
